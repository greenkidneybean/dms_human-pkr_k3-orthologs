```python
# plot parameters
plt.rcParams['font.family'] = 'Arial'

def custom_cmap(cmap_name, center, vmin, vmax):
    """Pulled from seaborn matrix.py to center cmap on WT value"""
    cmap = mpl.colormaps[cmap_name]
    under = cmap(-np.inf)
    over = cmap(np.inf)
    under_set = under != cmap(0)
    over_set = over != cmap(cmap.N - 1)

    vrange = max(vmax - center, center - vmin)
    normlize = mpl.colors.Normalize(center - vrange, center + vrange)
    cmin, cmax = normlize([vmin, vmax])
    cc = np.linspace(cmin, cmax, 256)
    cmap = mpl.colors.ListedColormap(cmap(cc))
    cmap.set_bad(color='gainsboro')
    if under_set:
        cmap.set_under(under)
    if over_set:
        cmap.set_over(over)

    return cmap


def parse_pdb_secondary_structure(pdb_path, chain_id):
    """
    Parse HELIX and SHEET records from a PDB file for a specific chain.

    PDB HELIX record (fixed columns, 1-indexed per PDB spec):
        cols 20       -> initial chain ID
        cols 22-25    -> initial residue seqnum (integer)
        cols 32       -> terminal chain ID
        cols 34-37    -> terminal residue seqnum (integer)

    PDB SHEET record:
        cols 22       -> initial chain ID
        cols 23-26    -> initial residue seqnum
        cols 33       -> terminal chain ID
        cols 34-37    -> terminal residue seqnum

    Returns a tuple (helices, strands) where each is a list of (start, end)
    inclusive integer residue ranges on the requested chain.
    """
    helices = []
    strands = []
    with open(pdb_path, 'r') as fh:
        for line in fh:
            rec = line[0:6]
            if rec == 'HELIX ':
                init_chain = line[19:20].strip()
                term_chain = line[31:32].strip()
                if init_chain == chain_id == term_chain:
                    try:
                        start = int(line[21:25].strip())
                        end = int(line[33:37].strip())
                        helices.append((start, end))
                    except ValueError:
                        continue
            elif rec == 'SHEET ':
                init_chain = line[21:22].strip()
                term_chain = line[32:33].strip()
                if init_chain == chain_id == term_chain:
                    try:
                        start = int(line[22:26].strip())
                        end = int(line[33:37].strip())
                        strands.append((start, end))
                    except ValueError:
                        continue
    return helices, strands


# new Roth-inspired AA order 250916MC
aa_list = ['H', 'K', 'R', 'D', 'E', 'N', 'Q', 'S', 'T', 'A', 'I', 'L', 'M', 'V', 'F', 'Y', 'W', 'C', 'G', 'P', '*']

aa_no_stops = aa_list.copy()
aa_no_stops.remove("*")

k3_conditions = ['K3Δ58', 'VACV K3', 'VARV C3', 'TPV K3', 'MYXV M156R', 'RCVZ vIF2α']
k3_names = ['K3Δ58', 'Vaccinia K3', 'Variola K3', 'Tanapox K3', 'Myxoma K3', 'Ranavirus K3']

# Define the 4 PKR structural regions (inclusive ranges)
pkr_sections = [(255, 278), (371, 385), (448, 455), (480, 506)]

# Sentinel column labels for the white gaps between sections
# (using negative ints that can't collide with real sites, and keeping them sortable per-section)
GAP_COLS = ['__gap1__', '__gap2__', '__gap3__']
# Sentinel row label for the white gap above the summary rows
GAP_ROW = '__row_gap__'

# ----------------------------------------------------------------------------
# TOGGLE 1: show WT sequence letters above the heatmap
# ----------------------------------------------------------------------------
SHOW_WT_SEQUENCE = True
WT_SEQ_LABEL = 'PKR WT Sequence'   # right-aligned label shown to the left of the WT letter row
WT_SEQ_LABEL_FONT_SIZE = 8         # pt (matches contact track label styling)
WT_SEQ_FONT_SIZE = 6               # pt, size of the individual WT letters
WT_SEQ_FONT_FAMILY = 'Arial'

# ----------------------------------------------------------------------------
# TOGGLE 2: contact / annotation triangle tracks shown above the heatmap.
#
# Set SHOW_CONTACT_TRACKS = True and fill CONTACT_TRACKS with one dict per
# track. Tracks are drawn top-to-bottom in the order given (i.e. the first
# entry is the top-most track, farthest from the heatmap).
#
# Each track dict supports:
#   'label' : str                 -> label text (8 pt Arial, right-aligned to heatmap)
#   'sites' : list[int]           -> PKR residue numbers where triangles appear
#   'color' : str (hex, e.g. '#000000') -> triangle fill color
# ----------------------------------------------------------------------------
SHOW_CONTACT_TRACKS = True

CONTACT_TRACK_DICT = {
  "eIF2a":{
      "label":"eIF2α Contact", 
      "color":"gray", 
      "sites":[274, 275, 276, 279, 335, 337, 338, 339, 340, 341, 342, 379, 382, 451, 452, 453, 483, 486, 487, 488, 489, 490, 491, 492, 493]
    },
  "vacv":{
    "label":"Vaccinia K3 Contact", 
    "color":"#786BAC", 
    "sites":[278, 339, 375, 379, 382, 414, 416, 435, 448, 449, 450, 451, 452, 453, 455, 460, 486, 487, 488, 489, 490, 492, 493, 496]
    },
  "varv":{
    "label":"Variola K3 Contact", 
    "color":"#E86562", 
    "sites":[278, 339, 375, 379, 382, 414, 416, 435, 448, 449, 450, 451, 452, 453, 455, 460, 486, 487, 488, 489, 490, 492, 493, 496]
    },
  "tpv":{
    "label":"Tanapox K3 Contact", 
    "color":"#F4B681", 
    "sites":[276, 277, 278, 375, 379, 382, 414, 416, 418, 435, 448, 450, 451, 452, 453, 454, 455, 480, 486, 487, 488, 489, 490, 491, 492, 493, 496]
    },
  "myxv":{
    "label":"Myxoma K3 Contact", 
    "color":"#6BB3FC", 
    "sites":[275, 277, 278, 304, 335, 337, 339, 375, 376, 414, 416, 434, 435, 448, 450, 451, 452, 453, 454, 455, 486, 487, 488, 489, 490, 492]
    },
  "rcv":{
    "label":"Ranavirus vIF2α Contact", 
    "color":"#A8D39B", 
    "sites":[275, 276, 277, 336, 337, 338, 339, 342, 375, 379, 382, 416, 451, 452, 453, 454, 483, 486, 487, 488, 489, 490, 492, 493, 495, 496, 499]
    },
  "conserved":{
    "label":"Conserved Site", 
    "color":"black", 
    "sites":[263, 267, 276, 277, 278, 279, 281, 283, 296, 298, 308, 309, 312, 315, 317, 319, 320, 323, 327, 362, 364, 365, 366, 367, 369, 374, 377, 397, 401, 404, 406, 407, 410, 411, 412, 413, 414, 415, 416, 417, 419, 420, 429, 430, 431, 432, 433, 437, 444, 445, 446, 450, 451, 454, 455, 457, 458, 459, 465, 469, 470, 474, 475, 476, 477, 480, 490, 511, 519, 523, 526]
    },
  "positive_vert":{
    "label":"Positive Selection", 
    "color":"red", 
    "sites":[261, 269, 270, 271, 272, 307, 314, 322, 360, 368, 375, 378, 379, 382, 385, 389, 394, 405, 428, 448, 449, 462, 471, 483, 486, 488, 491, 493, 500, 502, 504, 505, 514, 520, 524]
    }
}

CONTACT_TRACK_LIST = {
  "K3Δ58":["conserved","positive_vert","vacv","eIF2a"],
  "VACV K3":["conserved","positive_vert","vacv","eIF2a"],
  "VARV C3":["conserved","positive_vert",≈"eIF2a"],
  "TPV K3":["conserved","positive_vert","tpv","eIF2a"],
  "MYXV M156R":["conserved","positive_vert","myxv","eIF2a"],
  "RCVZ vIF2α":["conserved","positive_vert","rcv","eIF2a"],
  "z-score":[["conserved","positive_vert","rcv","tpv","varv","vacv","eIF2a"],
}

CONTACT_LABEL_FONT_SIZE = 8        # pt
CONTACT_TRACK_HEIGHT = 1.0         # in data (row) units per track; equilateral height is ~0.866
CONTACT_TRACK_SPACING_PT = 3.2     # blank vertical gap between adjacent tracks, in POINTS
                                   # (also used as the gap between WT sequence row and first contact track)

# ----------------------------------------------------------------------------
# TOGGLE 3: secondary-structure track derived from a PDB file.
#
# When enabled, one track is drawn above the heatmap (below the WT sequence row
# if that's also on). For each of the 4 heatmap sections, a colored rectangle
# spans the section's residue range; helices are drawn as white sine-wave
# squiggles and beta-strands as white arrows, with residue ranges pulled from
# HELIX and SHEET records of the specified PDB chain.
#
# Residue numbering in HELIX/SHEET records is assumed to match PKR residue
# numbering used throughout the heatmap.
# ----------------------------------------------------------------------------
SHOW_SECONDARY_STRUCTURE = True
PDB_PATH = '../../data/2A1A.pdb'          # path to local PDB file
PDB_CHAIN = 'B'                # which chain to use
SS_TRACK_HEIGHT = 1.2          # height of the secondary-structure track, in row units
SS_GAP_TO_WT = 0.1             # gap between SS track and WT letter row (if shown), in row units
SS_GAP_TO_HEATMAP = 0.05       # gap between SS track and heatmap top, in row units
# Colors per section (in the same order as pkr_sections)
SS_SECTION_COLORS = ['#E2B040', '#D46D2B', '#C3467E', '#6E2A59']
# Glyph styling (all glyphs are white, on the colored section background)
SS_GLYPH_COLOR = 'white'
SS_HELIX_AMPLITUDE = 0.28      # sine amplitude (row units); track half-height is 0.6, so stays inside
SS_HELIX_PERIOD = 2.0          # residues per sine period (tune for squiggle density)
SS_HELIX_LINEWIDTH = 1.3       # pt
SS_STRAND_THICKNESS = 0.35     # thickness of beta-strand shaft (row units)
SS_STRAND_HEAD_LENGTH = 1.0    # arrowhead length in residues (cell widths)
SS_STRAND_HEAD_WIDTH = 0.75    # arrowhead half-width (row units); total head width = 2x

for k3, k3_name in zip(k3_conditions, k3_names):
    k3_df = df.query('k3 == @k3')

    width_in = 8
    height_in = 4
    stroke = .25
    font = 'Arial'
    font_size = 7
    cmap_color = 'seismic_r'

    # custom cmap for k3-condition centered on PKR-WT
    auc_center = k3_df.query('pkr == "PKR-WT"').iloc[0]['auc_mean']
    auc_min = k3_df['auc_mean'].min()
    auc_max = k3_df['auc_mean'].max()
    centered_cmap = custom_cmap(cmap_color, auc_center, auc_min, auc_max)
    # Force masked (NaN) cells to WHITE for the gaps; they will blend with transparent bg on save,
    # but render as white on screen which matches the requested look.
    # We'll handle "not tested" gray cells separately by pre-filling them (they're already NaN -> gainsboro).
    # To keep "not tested" gray AND gaps white, we instead draw gaps as explicit white patches after plotting.

    # Build the full site list across all 4 sections with gap sentinels inserted between them
    ordered_cols = []
    for i, (lo, hi) in enumerate(pkr_sections):
        ordered_cols.extend(list(range(lo, hi + 1)))
        if i < len(pkr_sections) - 1:
            ordered_cols.append(GAP_COLS[i])

    # Filter data to the union of all region sites
    all_region_sites = [c for c in ordered_cols if isinstance(c, int)]
    temp_df = k3_df[k3_df['site'].isin(all_region_sites)].copy()
    temp_df['site'] = temp_df['site'].astype(int)

    # remove wt
    region_df = temp_df.query('pkr != "PKR-WT"')

    # --- WT black-circle dataframe ---
    cols = ['site', 'wt_aa']
    wt_df = region_df[cols].drop_duplicates()
    wt_df['site'] = wt_df['site'].astype(int)
    wt_df = wt_df.sort_values('site').set_index('site')
    wt_df = pd.get_dummies(wt_df['wt_aa'])
    missing_wt = list(set(aa_list) - set(wt_df.columns.tolist()))
    for aa in missing_wt:
        wt_df[aa] = False
    wt_df = wt_df[aa_list]
    wt_df.replace({True: 1, False: np.nan}, inplace=True)
    wt_df['Mean Missense'] = np.nan
    wt_df['Max Missense'] = np.nan
    wt_df['Min Missense'] = np.nan
    wt_df = wt_df.T  # rows = AAs + summary labels, cols = sites

    # Insert gap columns (all NaN) between sections
    for g in GAP_COLS:
        wt_df[g] = np.nan

    # Reorder columns to match section layout
    wt_df = wt_df[ordered_cols]

    # Insert the white spacer ROW above the three summary rows
    # Row order: all aa_list rows, then GAP_ROW, then the three summary rows
    summary_rows = ['Mean Missense', 'Max Missense', 'Min Missense']
    main_rows = [r for r in wt_df.index if r not in summary_rows]
    wt_df.loc[GAP_ROW] = np.nan
    wt_df = wt_df.loc[main_rows + [GAP_ROW] + summary_rows]
    wt_array = wt_df.values

    # --- Main data dataframe ---
    region_df = region_df.pivot(index='site', columns='var_aa', values='auc_mean')
    region_df = region_df[aa_list]
    region_df['Mean Missense'] = region_df[aa_no_stops].mean(axis=1)
    region_df['Max Missense'] = region_df[aa_no_stops].max(axis=1)
    region_df['Min Missense'] = region_df[aa_no_stops].min(axis=1)
    region_df = region_df.T  # rows = AAs + summary, cols = sites

    # Insert gap columns
    for g in GAP_COLS:
        region_df[g] = np.nan

    # Reorder columns
    region_df = region_df[ordered_cols]

    # Insert gap row above summary rows
    region_df.loc[GAP_ROW] = np.nan
    region_df = region_df.loc[main_rows + [GAP_ROW] + summary_rows]

    # --- Plot ---
    font = {'family': font, 'size': font_size}
    plt.figure(figsize=(width_in, height_in))

    min_value = region_df.min().min()
    max_value = region_df.max().max()
    even_ticks = np.arange(np.ceil(min_value), np.floor(max_value) + 1, 1)

    cbar_kws = {
        'aspect': 15,
        'orientation': 'vertical',
        'fraction': .05,
        'pad': .05,
        'shrink': .5,
        'anchor': (-.5, .5),
        'ticks': even_ticks,
    }

    g = sns.heatmap(
        data=region_df,
        cmap=centered_cmap,
        vmin=auc_min,
        vmax=auc_max,
        square=True,
        cbar_kws=cbar_kws,
    )

    # --- Overlay WHITE rectangles on the gap columns & gap row so they look like empty spacing ---
    # (Heatmap cells are 1x1 with corners at integer coords; cell (i,j) spans [j, j+1] x [i, i+1])
    col_labels = list(region_df.columns)
    row_labels = list(region_df.index)

    gap_col_idx = [i for i, c in enumerate(col_labels) if c in GAP_COLS]
    gap_row_idx = [i for i, r in enumerate(row_labels) if r == GAP_ROW]

    n_rows = len(row_labels)
    n_cols = len(col_labels)

    # White-out gap columns across full height
    for ci in gap_col_idx:
        g.add_patch(plt.Rectangle((ci, 0), 1, n_rows, facecolor='white',
                                  edgecolor='none', zorder=3, clip_on=False))
    # White-out gap row across full width
    for ri in gap_row_idx:
        g.add_patch(plt.Rectangle((0, ri), n_cols, 1, facecolor='white',
                                  edgecolor='none', zorder=3, clip_on=False))

    # --- WT black circles (skip gap rows/cols automatically since they're NaN) ---
    for i in range(len(wt_array)):
        for j in range(len(wt_array[i])):
            if np.isnan(wt_array[i, j]):
                continue
            if wt_array[i, j] == 1:
                circle = plt.Circle((j + 0.5, i + 0.5), 0.3, color='black', fill=True, zorder=4)
                g.add_artist(circle)

    # --- Colorbar ---
    cbar = g.collections[0].colorbar
    cbar.ax.tick_params(labelsize=font_size)
    cbar.set_label(f"PKR Functional Score versus {k3_name}", rotation=90,
                   labelpad=1, font=font, size=font_size)
    cbar.ax.yaxis.set_label_position('left')

    # --- Labels / ticks ---
    plt.xticks(fontproperties=font, fontsize=font_size)
    plt.yticks(fontproperties=font, fontsize=font_size)

    g.set_ylabel('')
    g.set_xlabel('PKR Residue', font=font, size=font_size)

    # Y tick labels: show AA labels and summary labels, hide the gap row label
    y_display_labels = ['' if r == GAP_ROW else r for r in row_labels]
    g.set_yticks(np.arange(len(row_labels)) + 0.5)
    g.set_yticklabels(y_display_labels, rotation=0, fontproperties=font)

    plt.tick_params(axis='both', which='both', width=stroke)

    # Alternating tick length + alternating horizontal offset for AA rows.
    # Odd-numbered labels (1st, 3rd, 5th... -> H, R, E, Q, T, L, V, Y, G) shift 2 pt LEFT.
    # Even-numbered labels (2nd, 4th, 6th... -> K, D, N, S, A, M, F, W, P) shift 3 pt RIGHT.
    # Summary rows: Mean Missense -> 2 pt LEFT, Max Missense -> 3 pt RIGHT, Min Missense -> 2 pt LEFT.
    from matplotlib.transforms import offset_copy
    fig = plt.gcf()
    ax = plt.gca()

    summary_offsets = {
        'Mean Missense': -2,
        'Max Missense': 3,
        'Min Missense': -2,
    }

    major_ticks = ax.yaxis.get_major_ticks()
    aa_counter = 0  # counts only AA rows (skips gap + summary rows)
    for i, tick in enumerate(major_ticks):
        row_label = row_labels[i]
        if row_label == GAP_ROW:
            tick.tick1line.set_markersize(0)
            continue
        # Tick mark length alternation (preserve original behavior for non-gap rows)
        tick_length = 8 if i % 2 == 0 else 3
        tick.tick1line.set_markersize(tick_length)

        # Horizontal offset: AA rows alternate, summary rows have explicit shifts
        dx_points = None
        if row_label in aa_list:
            dx_points = -2 if aa_counter % 2 == 0 else 3
            aa_counter += 1
        elif row_label in summary_offsets:
            dx_points = summary_offsets[row_label]

        if dx_points is not None:
            label = tick.label1
            label.set_transform(
                offset_copy(label.get_transform(), fig=fig, x=dx_points, y=0, units='points')
            )

    # X tick labels: major ticks every 5 residues, skipping gap columns
    x_index = []
    x_value = []
    for i, x in enumerate(col_labels):
        if isinstance(x, int) and x % 5 == 0:
            x_index.append(i + 0.5)
            x_value.append(x)
    g.set_xticks(x_index)
    g.set_xticklabels(x_value, fontproperties=font, rotation=45)

    # Minor ticks on every non-gap column
    minor_positions = [i + 0.5 for i, c in enumerate(col_labels) if isinstance(c, int)]
    g.set_xticks(minor_positions, minor=True)

    # --- Section outlines ---
    # Turn off default spines so no thin lines span across the gap columns/row
    for spine in g.spines.values():
        spine.set_visible(False)

    # Find column index spans for each of the 4 sections (contiguous runs of real int columns)
    section_spans = []  # list of (start_col_idx, end_col_idx) inclusive
    current_start = None
    current_end = None
    for i, c in enumerate(col_labels):
        if isinstance(c, int):
            if current_start is None:
                current_start = i
            current_end = i
        else:
            if current_start is not None:
                section_spans.append((current_start, current_end))
                current_start = None
    if current_start is not None:
        section_spans.append((current_start, current_end))

    # Identify the two row-blocks: AA block (above the gap row) and summary block (below)
    gap_row_position = next(i for i, r in enumerate(row_labels) if r == GAP_ROW)
    row_blocks = [
        (0, gap_row_position - 1),                     # AA block (H..*)
        (gap_row_position + 1, len(row_labels) - 1),   # summary block (Mean/Max/Min Missense)
    ]

    # Draw a 0.2 pt black rectangle around each (section x row-block) combination.
    # This produces 8 separate outlines with no connecting strokes across the gaps.
    section_stroke = 0.2  # in points, as requested
    for (c_start, c_end) in section_spans:
        for (r_start, r_end) in row_blocks:
            rect = plt.Rectangle(
                (c_start, r_start),
                (c_end - c_start + 1),
                (r_end - r_start + 1),
                fill=False,
                edgecolor='black',
                linewidth=section_stroke,
                zorder=5,
                clip_on=False,
            )
            g.add_patch(rect)

    # ------------------------------------------------------------------
    # Helpers for anything drawn ABOVE the heatmap
    # ------------------------------------------------------------------
    # Map: PKR site (int) -> column index in the heatmap (x center is idx + 0.5)
    site_to_colidx = {c: i for i, c in enumerate(col_labels) if isinstance(c, int)}
    # Map: PKR site -> WT AA (pulled from the original k3_df, which has wt_aa per site)
    site_to_wt = (
        k3_df[['site', 'wt_aa']]
        .drop_duplicates()
        .dropna()
        .assign(site=lambda d: d['site'].astype(int))
        .set_index('site')['wt_aa']
        .to_dict()
    )

    # Convert points -> row-unit height. 1 data unit in y == the pixel-height of
    # one row; 1 pt == fig.dpi / 72 pixels. So 1 pt in row units depends on the
    # current axes transform.
    def _points_to_data_y(pts):
        # Transform (0, 0) and (0, pts) through the inverse display->data transform.
        trans = ax.transData
        inv = trans.inverted()
        p0 = ax.transAxes.transform((0, 0))  # just to get a fig ref
        # Use a direct offset in display pixels: pts * dpi / 72 pixels
        dpi = fig.get_dpi()
        px = pts * dpi / 72.0
        # Pick any display point and shift up by px pixels, then convert delta
        d0 = trans.transform((0, 0))
        d1 = (d0[0], d0[1] + px)
        y0 = inv.transform(d0)[1]
        y1 = inv.transform(d1)[1]
        return abs(y1 - y0)

    # Track the running top of content drawn above the heatmap (in data/row units).
    # The heatmap's top edge is y=0; anything above has y<0.
    content_top_y = 0.0

    # ------------------------------------------------------------------
    # TOGGLE 3: secondary-structure track (closest to heatmap)
    # ------------------------------------------------------------------
    if SHOW_SECONDARY_STRUCTURE:
        helices, strands = parse_pdb_secondary_structure(PDB_PATH, PDB_CHAIN)

        # Small gap between heatmap top and the SS track
        content_top_y -= SS_GAP_TO_HEATMAP

        ss_top_y = content_top_y - SS_TRACK_HEIGHT
        ss_mid_y = (content_top_y + ss_top_y) / 2.0

        # --- Colored background rectangles, one per section ---
        for (lo, hi), color, (c_start, c_end) in zip(
            pkr_sections, SS_SECTION_COLORS, section_spans
        ):
            rect = plt.Rectangle(
                (c_start, ss_top_y),
                (c_end - c_start + 1),
                SS_TRACK_HEIGHT,
                facecolor=color,
                edgecolor='none',
                zorder=5,
                clip_on=False,
            )
            g.add_patch(rect)

        # Helper: which (section_idx, residue_low, residue_high) does a range fall in?
        # Helices/strands can span a section; we clip them to each section they touch.
        def _clip_to_sections(start, end):
            """Yield (c_start_idx, c_end_idx, res_start, res_end) per section the range intersects."""
            for (lo, hi), (c_start, c_end) in zip(pkr_sections, section_spans):
                if end < lo or start > hi:
                    continue
                r0 = max(start, lo)
                r1 = min(end, hi)
                yield c_start, c_end, r0, r1

        # --- Draw helices as white sine squiggles ---
        for (h_start, h_end) in helices:
            for c_start, c_end, r0, r1 in _clip_to_sections(h_start, h_end):
                # x-range in data coords: center of starting cell to center of ending cell
                # Site r is at column (c_start + (r - pkr_sections[i][0])), center at +0.5
                section_idx = next(i for i, (lo, hi) in enumerate(pkr_sections) if lo <= r0 <= hi)
                section_lo = pkr_sections[section_idx][0]
                x0 = c_start + (r0 - section_lo) + 0.5
                x1 = c_start + (r1 - section_lo) + 0.5
                if x1 <= x0:
                    continue
                n_samples = max(int((x1 - x0) * 25), 10)
                xs = np.linspace(x0, x1, n_samples)
                ys = ss_mid_y + SS_HELIX_AMPLITUDE * np.sin(
                    2 * np.pi * (xs - x0) / SS_HELIX_PERIOD
                )
                g.plot(
                    xs, ys,
                    color=SS_GLYPH_COLOR,
                    linewidth=SS_HELIX_LINEWIDTH,
                    solid_capstyle='round',
                    zorder=6,
                    clip_on=False,
                )

        # --- Draw beta strands as white arrows ---
        for (s_start, s_end) in strands:
            for c_start, c_end, r0, r1 in _clip_to_sections(s_start, s_end):
                section_idx = next(i for i, (lo, hi) in enumerate(pkr_sections) if lo <= r0 <= hi)
                section_lo = pkr_sections[section_idx][0]
                # Strand spans from left edge of first residue cell to right edge of last cell
                x0 = c_start + (r0 - section_lo)
                x1 = c_start + (r1 - section_lo) + 1.0
                if x1 <= x0:
                    continue

                # Shaft is a rectangle; arrowhead is a triangle at the C-terminal end (x1)
                shaft_half = SS_STRAND_THICKNESS / 2.0
                head_len = min(SS_STRAND_HEAD_LENGTH, (x1 - x0) * 0.6)
                head_half_width = SS_STRAND_HEAD_WIDTH

                # Build a single polygon for the whole arrow shape
                arrow_pts = [
                    (x0,                ss_mid_y - shaft_half),   # bottom-left of shaft
                    (x1 - head_len,     ss_mid_y - shaft_half),   # bottom-right of shaft
                    (x1 - head_len,     ss_mid_y - head_half_width),  # arrowhead bottom flange
                    (x1,                ss_mid_y),                # arrowhead tip
                    (x1 - head_len,     ss_mid_y + head_half_width),  # arrowhead top flange
                    (x1 - head_len,     ss_mid_y + shaft_half),   # top-right of shaft
                    (x0,                ss_mid_y + shaft_half),   # top-left of shaft
                ]
                arrow = plt.Polygon(
                    arrow_pts,
                    closed=True,
                    facecolor=SS_GLYPH_COLOR,
                    edgecolor='none',
                    zorder=6,
                    clip_on=False,
                )
                g.add_patch(arrow)

        content_top_y = ss_top_y  # move ceiling up past SS track

    # ------------------------------------------------------------------
    # TOGGLE 1: WT sequence letters above the SS track (or heatmap)
    # ------------------------------------------------------------------
    if SHOW_WT_SEQUENCE:
        # If SS track is shown, add a small gap before WT letters
        if SHOW_SECONDARY_STRUCTURE:
            content_top_y -= SS_GAP_TO_WT

        wt_row_height = 1.0  # one cell-height row, same as heatmap cells
        wt_row_top = content_top_y - wt_row_height
        wt_row_center_y = wt_row_top + wt_row_height / 2.0

        for site, col_idx in site_to_colidx.items():
            wt_aa = site_to_wt.get(site)
            if wt_aa is None:
                continue
            g.text(
                col_idx + 0.5,
                wt_row_center_y,
                str(wt_aa),
                ha='center',
                va='center',
                fontsize=WT_SEQ_FONT_SIZE,
                family=WT_SEQ_FONT_FAMILY,
                color='black',
                clip_on=False,
                zorder=6,
            )

        # --- Right-aligned label for the WT sequence row (matches contact track label styling) ---
        if WT_SEQ_LABEL:
            g.text(
                -0.3,                  # same x-offset as contact track labels
                wt_row_center_y,
                WT_SEQ_LABEL,
                ha='right',
                va='center',
                fontsize=WT_SEQ_LABEL_FONT_SIZE,
                family='Arial',
                color='black',
                clip_on=False,
                zorder=6,
            )

        content_top_y = wt_row_top  # move the ceiling up

    # ------------------------------------------------------------------
    # TOGGLE 2: contact / annotation triangle tracks (topmost)
    # ------------------------------------------------------------------
    ### START HERE
    CONTACT_TRACKS = [CONTACT_TRACK_DICT[a] for a in CONTACT_TRACK_LIST[k3]]
    
    if SHOW_CONTACT_TRACKS and len(CONTACT_TRACKS) > 0:
        # Equilateral triangle with base = 1 cell width -> height = sqrt(3)/2
        tri_height = (3 ** 0.5) / 2.0

        # Convert the points-based inter-track spacing to row units NOW (after
        # axes have been drawn and sizing is stable).
        contact_track_spacing = _points_to_data_y(CONTACT_TRACK_SPACING_PT)

        # Gap between the WT sequence row (or heatmap top) and the FIRST (bottom-most)
        # contact track uses the same point-based spacing as the gap BETWEEN adjacent
        # contact tracks -- so the WT label sits the same visual distance from the
        # lowest contact label as the contact labels sit from each other.
        content_top_y -= contact_track_spacing

        # Tracks are drawn bottom-up in data coords (negative y), but listed
        # top-to-bottom in CONTACT_TRACKS. So iterate reversed so the first
        # entry ends up farthest from the heatmap.
        for track in reversed(CONTACT_TRACKS):
            label = track.get('label', '')
            sites = track.get('sites', [])
            color = track.get('color', '#000000')

            # Vertical band for this track: tip (pointing down) at content_top_y,
            # base (top of triangle) at content_top_y - tri_height
            tip_y = content_top_y
            base_y = content_top_y - tri_height

            # Draw a triangle for each listed site (skip sites that aren't in the map)
            for site in sites:
                col_idx = site_to_colidx.get(int(site))
                if col_idx is None:
                    continue
                x_center = col_idx + 0.5
                triangle = plt.Polygon(
                    [
                        (x_center, tip_y),                 # bottom tip, centered on cell
                        (x_center - 0.5, base_y),          # top-left vertex
                        (x_center + 0.5, base_y),          # top-right vertex
                    ],
                    closed=True,
                    facecolor=color,
                    edgecolor='none',
                    linewidth=0,
                    zorder=6,
                    clip_on=False,
                )
                g.add_patch(triangle)

            # Label, 8 pt Arial, right-aligned so its right edge sits just left
            # of the heatmap's left edge (x = 0 in data coords).
            label_y = (tip_y + base_y) / 2.0
            g.text(
                -0.3,                 # small data-unit offset; fine-tunes to the left of x=0
                label_y,
                label,
                ha='right',
                va='center',
                fontsize=CONTACT_LABEL_FONT_SIZE,
                family='Arial',
                color='black',
                clip_on=False,
                zorder=6,
            )

            # Advance the ceiling upward: this track's top + spacing
            content_top_y = base_y - contact_track_spacing

    # Extend the y-axis upward so annotations above the heatmap aren't clipped on screen.
    # (bbox_inches='tight' will still crop to artists, so this doesn't add whitespace on save.)
    if content_top_y < 0:
        current_ylim = g.get_ylim()  # (bottom, top) in data coords; top is smaller / more negative
        g.set_ylim(current_ylim[0], min(current_ylim[1], content_top_y - 0.5))

    # save plot
    plot_name = f'new-heatmap_{k3}.svg'
    output_file = os.path.join(save_dir, plot_name)
    plt.savefig(output_file, bbox_inches='tight', format='svg', transparent=True)

    plt.show()

    # --- STRUCTURE COLORING (unchanged logic; operates on numeric sites only) ---
    def color_structure(score_category='Mean Missense', depiction='surface'):
        # Get values only for real (numeric) site columns
        numeric_cols = [c for c in region_df.columns if isinstance(c, int)]
        values = region_df.loc[score_category, numeric_cols].values
        values = values[~np.isnan(values)]
        norm = mcolors.Normalize(vmin=auc_min, vmax=auc_max)
        colors = [centered_cmap(norm(value)) for value in values.flatten()]
        hex_colors = [mcolors.rgb2hex(color) for color in colors]
        pkr_sites = sorted(list(set(df.query('pkr != "PKR-WT"')['site'].astype(int).tolist())))

        cmd_list = []
        for site, color in zip(pkr_sites, hex_colors):
            cmd_list.append(f"color #1/A:{site} {color} {depiction};")

        print(f'# ChimeraX Coloring: {k3}, Scoring Condition: {score_category}')
        print(''.join([f'color #1/A #6dc091 {depiction};'] + cmd_list))
        print()

    color_structure('Mean Missense', 'cartoon')
    color_structure('Min Missense', 'cartoon')
    color_structure('Max Missense', 'cartoon')
    color_structure('Mean Missense', 'surface')
    color_structure('Min Missense', 'surface')
    color_structure('Max Missense', 'surface')
```
