```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['pdf.fonttype'] = 42 # make text editable in .pdf
import numpy as np
```

```python
# Read the .pim file
input_file = 'data/pim_eif2a_k2.csv'

names = ['RCVZ_Ranavirus','Human_eIF2a', 'EPTV_Eptesipox_virus','MYXV_Myxoma_virus','RFV_Rabbit_fibroma_virus','YKV_Yokapox_virus','COTV_Cotia_virus','MPXV_longsub_Monkeypox_virus','VACV_Vaccinia_virus','VARV_Variola_virus','TATP_Taterapox_virus','CMLVM96_Camelpox_virus','RCPV_Racoonpox_virus','SKPV_Skunkpox_virus','VPV_Volepox_virus','SWPV_Swinepox_virus','TPV_Tanapox_virus', 'YMTV_Yaba_monkey_tumor_virus','SPPV_Sheeppox_virus','DPV_Deerpox_virus']
new_names = [
 'Ranavirus',
 'Human eIF2α',
 'Eptesipox',
 'Myxoma',
 'Rabbit fibroma',
 'Yokapox',
 'Cotia',
 'Monkeypox',
 'Vaccinia',
 'Variola',
 'Taterapox',
 'Camelpox',
 'Racoonpox',
 'Skunkpox',
 'Volepox',
 'Swinepox',
 'Tanapox',
 'Yaba monkey tumor',
 'Sheeppox',
 'Deerpox'
]

accession_names = [
 'Ranavirus AAY86037.1',
 'Human eIF2α NP_001127006.1',
 'Eptesipox YP_009407963.1',
 'Myxoma virus NP_051869.1',
 'Rabbit fibroma NP_051897.1',
 'Yokapox YP_004821369.1',
 'Cotia YP_005296217.1',
 'Monkeypox PP556318.1',
 'Vaccinia YP_232916.1',
 'Variola NP_042067.1',
 'Taterapox YP_717344.1',
 'Camelpox NP_570422.1',
 'Racoonpox YP_009143343.1',
 'Skunkpox YP_009282727.1',
 'Volepox YP_009281781.1',
 'Swinepox NP_570170.1',
 'Tanapox AAD46178.1',
 'Yaba monkey tumor NP_938270.1',
 'Sheeppox NP_659587.1',
 'Deerpox ABI99007.1'
]

df = pd.read_csv(input_file, names=new_names)
df.index = pd.Series(new_names)
df.head()
```

```python
#fig, ax = plt.subplots(figsize=(5, 5), dpi=300) 
font = {'family': 'Arial', 'size': 8}
plt.rc('font', **font)

plt.rc('lines', linewidth=.75)

plt.figure(figsize=(4.5, 4.5)) # save size

ax = sns.heatmap(
    data=df, 
    yticklabels=new_names, 
    cmap='viridis', 
    square=True,
    cbar_kws={'label': 'K3 Ortholog Sequence Percent Identity'}  # Add label to color bar
)

# Get the x-axis labels
labels = ax.get_xticklabels()

# Set the rotation of the x-axis labels
plt.setp(labels, rotation=45, ha='right')

# Customize tick marks stroke width
ax.tick_params(axis='both', which='major', width=.25, length=3)  # Adjust width and length


# Optionally adjust the x-axis label spacing
plt.subplots_adjust(bottom=0.3)  # Increase the bottom margin

fig = ax.get_figure()
#fig.savefig(f'img/k3_pim.png', bbox_inches='tight', format='png', dpi=300)
fig.savefig(f'img/pim_sm_k3-orthologs_names.svg', bbox_inches='tight', format='svg')

plt.show()

# Display the plot
plt.show()
```

```python
with open('data/k3-orthologs.pim', 'r') as file:
    lines = file.readlines()

# Parse the header and sequences
header = lines[0].strip().split()
sequences = [seq.strip().split() for seq in lines[1:]]

# Create a dictionary from the data
data = {header[0]: []}
for i in range(1, len(header)):
    data[header[i]] = [sequences[i-1][0]]
    data[header[0]].append(sequences[i-1][0])

for i, row in enumerate(sequences):
    for j, val in enumerate(row[1:]):
        data[header[j+1]][i+1] = float(val)

# Create the DataFrame
df = pd.DataFrame(data)

# Print the DataFrame
print(df)
```
