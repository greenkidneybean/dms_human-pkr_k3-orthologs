```python
#!/usr/bin/env python3
"""List conserved-column residue positions (*, :, .) from a ClustalW alignment,
numbered according to a chosen reference sequence.

Biopython's clustal parser stores the conservation line (the * : . row) in
alignment.column_annotations["clustal_consensus"], so we just walk the columns
alongside the reference row and count non-gap residues to get the numbering.
"""

from Bio import AlignIO

ALIGNMENT_FILE = "../../data/muscle-I20260805-172747-0287-69493240-p1m.aln-clustalw"
REFERENCE_ID = "PKR"   # residue numbering is based on this sequence

alignment = AlignIO.read(ALIGNMENT_FILE, "clustal")
conservation = alignment.column_annotations["clustal_consensus"]

# Reference row used for numbering.
ref = next(rec for rec in alignment if rec.id == REFERENCE_ID)

star, colon, period = [], [], []
pos = 0  # 1-based residue number in the ungapped reference sequence

for aa, symbol in zip(str(ref.seq), conservation):
    if aa == "-":
        continue          # gap in reference -> no residue number at this column
    pos += 1
    if symbol == "*":
        star.append(pos)
    elif symbol == ":":
        colon.append(pos)
    elif symbol == ".":
        period.append(pos)

fmt = lambda nums: ",".join(str(n) for n in nums)

print(f"*: {fmt(star)}\n")
print(f": {fmt(colon)}\n")
print(f".: {fmt(period)}")
```
